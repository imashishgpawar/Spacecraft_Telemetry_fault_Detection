```python
import pandas as pd

file_path = "/content/drive/MyDrive/Spacecraft_Project_Data/combined_spacecraft_telemetry.csv"

df = pd.read_csv(file_path, nrows=100000)
```

```python
X = df.drop(columns=['datetime'])

print(X.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled.shape)
```

```python
WINDOW = 60
```

```python
import numpy as np

def create_sequences(data, window=60):

    X = []

    for i in range(len(data) - window):
        X.append(data[i:i+window])

    return np.array(X)

X_seq = create_sequences(X_scaled, WINDOW)

print(X_seq.shape)
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, RepeatVector, TimeDistributed, Dense
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt

# ----------------------------
# Train-Test Split
# ----------------------------
X_train, X_test = train_test_split(
    X_seq,
    test_size=0.2,
    random_state=42
)

print("Train Shape:", X_train.shape)
print("Test Shape :", X_test.shape)

# ----------------------------
# LSTM Autoencoder
# ----------------------------
timesteps = X_train.shape[1]
features = X_train.shape[2]

inputs = Input(shape=(timesteps, features))

encoded = LSTM(64, activation='tanh')(inputs)

decoded = RepeatVector(timesteps)(encoded)

decoded = LSTM(
    64,
    activation='tanh',
    return_sequences=True
)(decoded)

outputs = TimeDistributed(
    Dense(features)
)(decoded)

model = Model(inputs, outputs)

model.compile(
    optimizer='adam',
    loss='mse'
)

model.summary()

# ----------------------------
# Train Model
# ----------------------------
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    X_train,
    epochs=10,
    batch_size=128,
    validation_data=(X_test, X_test),
    callbacks=[early_stop]
)

# ----------------------------
# Reconstruction Error
# ----------------------------
reconstructions = model.predict(X_test)

mse = np.mean(
    np.power(X_test - reconstructions, 2),
    axis=(1,2)
)

threshold = np.percentile(mse, 95)

predictions = (mse > threshold).astype(int)

print("\nThreshold:", threshold)
print("Detected Anomalies:", np.sum(predictions))

# ----------------------------
# Plot Error Distribution
# ----------------------------
plt.figure(figsize=(10,5))
plt.hist(mse, bins=50)
plt.axvline(
    threshold,
    linestyle='--'
)
plt.title("Reconstruction Error Distribution")
plt.xlabel("MSE")
plt.ylabel("Count")
plt.show()
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# =====================================================================
# STEP 1: PATH CONFIGURATION & OPTIMIZED MEMORY LOADING
# =====================================================================
# Update these paths if your files are located elsewhere in your Google Drive or PC
TELEMETRY_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/combined_spacecraft_telemetry.csv"
LABELS_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/labels.csv"
print("======= STEP 1: LOADING & ALIGNING TIMESTAMPS =======")

# Determine the structural column headers to enforce a memory-light float32 format
sample_read = pd.read_csv(TELEMETRY_FILE, nrows=2)
float32_dtypes = {col: np.float32 for col in sample_read.columns if col != 'datetime'}

# Load the core telemetry data
df_telemetry = pd.read_csv(
    TELEMETRY_FILE,
    index_col='datetime',
    parse_dates=True,
    dtype=float32_dtypes,
    low_memory=True,
    memory_map=True
)

# Crucial Fix: Localize the telemetry index to UTC to prevent naive/aware comparison crashes
if df_telemetry.index.tz is None:
    df_telemetry.index = df_telemetry.index.tz_localize('UTC')
else:
    df_telemetry.index = df_telemetry.index.tz_convert('UTC')

# Load the target annotation logs and synchronize time zones to UTC
df_labels = pd.read_csv(LABELS_FILE)
df_labels['StartTime'] = pd.to_datetime(df_labels['StartTime']).dt.tz_convert('UTC')
df_labels['EndTime'] = pd.to_datetime(df_labels['EndTime']).dt.tz_convert('UTC')

print(f"Loaded {df_telemetry.shape[0]} rows and {df_telemetry.shape[1]} features.")

# =====================================================================
# STEP 2: SUPERVISED STEP-BY-STEP TARGET WINDOW MAPPING
# =====================================================================
print("\n======= STEP 2: GENERATING TARGET LABELS (y) =======")

# Initialize array with nominal states (0)
y_continuous = np.zeros(len(df_telemetry), dtype=np.int32)

# Vectorized lookup mask scanning active fault indices
for _, row in df_labels.iterrows():
    mask = (df_telemetry.index >= row['StartTime']) & (df_telemetry.index <= row['EndTime'])
    y_continuous[mask] = 1

nominal_count = np.sum(y_continuous == 0)
anomalous_count = np.sum(y_continuous == 1)
print(f"Total Labels Created: Nominal: {nominal_count} | Anomalous: {anomalous_count}")
print(f"Dataset Anomaly Percentage: {(anomalous_count / len(y_continuous)) * 100:.4f}%")

# =====================================================================
# STEP 3: SUBSET DEVELOPMENT HANDLING (PREVENT RAM BLOWOUT)
# =====================================================================
print("\n======= STEP 3: CREATING SAMPLES AND SEQUENCES =======")

# Due to 7.3 Million steps, take a large representative sequence block (e.g., 200k steps)
# to train stably on a Colab/local system without running out of RAM memory.
SUBSET_ROWS = 200000
X_raw = df_telemetry.iloc[:SUBSET_ROWS].values
y_raw = y_continuous[:SUBSET_ROWS]

WINDOW_SIZE = 60

def rolling_window_generator(X_data, y_data, window):
    X_seq, y_seq = [], []
    for i in range(len(X_data) - window):
        X_seq.append(X_data[i : i + window])
        y_seq.append(y_data[i + window]) # Predict if the final step of the window is anomalous
    return np.array(X_seq, dtype=np.float32), np.array(y_seq, dtype=np.int32)

X_seq, y_seq = rolling_window_generator(X_raw, y_raw, WINDOW_SIZE)
print(f"Generated Sequence Tensors Shape: {X_seq.shape}")
print(f"Generated Label Shape: {y_seq.shape}")

# =====================================================================
# STEP 4: STRATIFIED TRAIN-TEST SPLIT
# =====================================================================
X_train, X_test, y_train, y_test = train_test_split(
    X_seq, y_seq,
    test_size=0.2,
    random_state=42,
    stratify=y_seq # Maintains equal normal/anomaly ratios in both sets
)

# =====================================================================
# STEP 5: MODEL 1 ARCHITECTURE - SUPERVISED LSTM CLASSIFIER
# =====================================================================
print("\n======= STEP 5: COMPILING SUPERVISED LSTM CLASSIFIER =======")

model = Sequential([
    LSTM(64, input_shape=(X_train.shape[1], X_train.shape[2]), return_sequences=False),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid') # Binary Output: 0 for Normal Heat, 1 for Directional / Thermal Anomaly
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

# =====================================================================
# STEP 6: MODEL TRAINING WITH EARLY STOPPING
# =====================================================================
print("\n======= STEP 6: TRAINING BASELINE LSTM CLASSIFIER =======")

early_stop = EarlyStopping(monitor='val_loss', patience=2, restore_best_weights=True)

history = model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=256,
    validation_data=(X_test, y_test),
    callbacks=[early_stop],
    verbose=1
)

# =====================================================================
# STEP 7: EVALUATION METRICS & GRAPH CURVES
# =====================================================================
print("\n======= STEP 7: EVALUATING SUPERVISED MODEL METRICS =======")

y_pred_probs = model.predict(X_test, batch_size=256)
y_pred = (y_pred_probs > 0.5).astype(int)

# Calculate true performance metrics
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, zero_division=0)
rec = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print("\n================ MODEL 1 PERFORMANCE REVEALED ================")
print(f"Classification Accuracy : {acc * 100:.2f}%")
print(f"Precision Performance   : {prec:.4f}")
print(f"Recall (Sensitivity)    : {rec:.4f}")
print(f"F1-Score Performance    : {f1:.4f}")
print("\nConfusion Matrix Layout:")
print(confusion_matrix(y_test, y_pred))
print("\nDetailed Summary:")
print(classification_report(y_test, y_pred))

# Plot and render loss curves
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Supervised LSTM Loss Trend')
plt.xlabel('Epochs')
plt.ylabel('Loss Value')
plt.legend()

# Plot accuracy curves
plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Supervised LSTM Accuracy Trend')
plt.xlabel('Epochs')
plt.ylabel('Accuracy Score')
plt.legend()
plt.tight_layout()
plt.show()

# Save the properly trained baseline model
model.save("Supervised_LSTM_Classification_Model.h5")
print("\nModel saved successfully as: Supervised_LSTM_Classification_Model.h5")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# =====================================================================
# STEP 1: DEFINE CONFIGURATIONS & SUBSYSTEM CHANNELS
# =====================================================================
TELEMETRY_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/combined_spacecraft_telemetry.csv"
LABELS_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/labels.csv"

# Isolating features as defined in the project engineering layout
thermal_channels = [f'channel_{i}' for i in list(range(12, 41)) + list(range(47, 53)) + list(range(57, 61))]
adcs_channels = [f'channel_{i}' for i in list(range(53, 57)) + list(range(70, 77))]
target_features = ['datetime'] + thermal_channels + adcs_channels

print("======= STEP 1: LOADING LABELS AND CONFIGURING STREAM =======")

# Load labels file first and force time zones to UTC
df_labels = pd.read_csv(LABELS_FILE)
df_labels['StartTime'] = pd.to_datetime(df_labels['StartTime']).dt.tz_convert('UTC')
df_labels['EndTime'] = pd.to_datetime(df_labels['EndTime']).dt.tz_convert('UTC')

# =====================================================================
# STEP 2: HIGH-EFFICIENCY CHUNKED LOADING (RAM PROTECTION)
# =====================================================================
# Instead of loading 7.3 million rows, we stream a highly informative subset
# (300,000 rows) using explicit chunks to keep memory consumption under 2 GB.
CHUNK_SIZE = 50000
MAX_ROWS = 300000

X_list = []
y_list = []
rows_processed = 0

print(f"Streaming data sequentially up to {MAX_ROWS} rows...")

for chunk in pd.read_csv(TELEMETRY_FILE, usecols=target_features, chunksize=CHUNK_SIZE):
    if rows_processed >= MAX_ROWS:
        break

    # Standardize timestamps in this chunk to UTC
    chunk['datetime'] = pd.to_datetime(chunk['datetime'])
    chunk.set_index('datetime', inplace=True)
    if chunk.index.tz is None:
        chunk.index = chunk.index.tz_localize('UTC')

    # Generate true binary labels for this chunk
    y_chunk = np.zeros(len(chunk), dtype=np.int32)
    for _, row in df_labels.iterrows():
        mask = (chunk.index >= row['StartTime']) & (chunk.index <= row['EndTime'])
        y_chunk[mask] = 1

    # Append values to lists as ultra-light matrices
    X_list.append(chunk.values.astype(np.float32))
    y_list.append(y_chunk)

    rows_processed += len(chunk)
    print(f"Processed: {rows_processed}/{MAX_ROWS} rows safely...")

# Combine chunks into unified baseline arrays
X_raw = np.vstack(X_list)
y_raw = np.concatenate(y_list)

# Explicit garbage collection to free system RAM instantly
del X_list, y_list
print(f"\nFinal Consolidated Raw Data Matrix: {X_raw.shape}")
print(f"Class Counts -> Nominal: {np.sum(y_raw==0)} | Anomalous: {np.sum(y_raw==1)}")

# =====================================================================
# STEP 3: CONSTRUCT TIME-SERIES WINDOW SEQUENCES
# =====================================================================
print("\n======= STEP 3: CREATING ROLLING SEQUENCES =======")
WINDOW_SIZE = 60

def rolling_window_generator(X_data, y_data, window):
    X_seq, y_seq = [], []
    for i in range(len(X_data) - window):
        X_seq.append(X_data[i : i + window])
        y_seq.append(y_data[i + window]) # Predict class at the end of the timeline
    return np.array(X_seq, dtype=np.float32), np.array(y_seq, dtype=np.int32)

X_seq, y_seq = rolling_window_generator(X_raw, y_raw, WINDOW_SIZE)
print(f"LSTM Trainable Input Tensor Shape: {X_seq.shape}")

# Clear raw variables from cache
del X_raw, y_raw

# =====================================================================
# STEP 4: STRATIFIED SPLITTING
# =====================================================================
X_train, X_test, y_train, y_test = train_test_split(
    X_seq, y_seq,
    test_size=0.2,
    random_state=42,
    stratify=y_seq if np.sum(y_seq == 1) > 1 else None
)

# =====================================================================
# STEP 5: MODEL COMPILATION (MODERN INPUT SYNTAX)
# =====================================================================
print("\n======= STEP 4: COMPILING SUPERVISED LSTM CLASSIFIER =======")

model = Sequential([
    Input(shape=(X_train.shape[1], X_train.shape[2])),
    LSTM(64, return_sequences=False),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid') # Probability mapping output
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

# =====================================================================
# STEP 6: TRAIN baseline CLASSIFIER
# =====================================================================
print("\n======= STEP 5: TRAINING BASELINE MODEL =======")
early_stop = EarlyStopping(monitor='val_loss', patience=2, restore_best_weights=True)

history = model.fit(
    X_train, y_train,
    epochs=10,
    batch_size=256,
    validation_data=(X_test, y_test),
    callbacks=[early_stop],
    verbose=1
)

# =====================================================================
# STEP 7: STRUCTURAL METRIC COMPUTATION & EVALUATION
# =====================================================================
print("\n======= STEP 6: EVALUATING SUPERVISED PERFORMANCE =======")
y_pred_probs = model.predict(X_test, batch_size=256)
y_pred = (y_pred_probs > 0.5).astype(int)

# Extract real metrics mapping
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, zero_division=0)
rec = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print("\n================ FINAL MODEL 1 METRICS ================")
print(f"Accuracy Performance : {acc * 100:.2f}%")
print(f"Precision Score      : {prec:.4f}")
print(f"Recall (Sensitivity) : {rec:.4f}")
print(f"F1-Score Performance : {f1:.4f}")
print("\nConfusion Matrix Output:")
print(confusion_matrix(y_test, y_pred))
print("\nDetailed Summary:")
print(classification_report(y_test, y_pred, zero_division=0))

# Render performance trends
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Loss Path Analysis')
plt.xlabel('Epochs')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.title('Accuracy Path Analysis')
plt.xlabel('Epochs')
plt.legend()
plt.show()

# Save using native format
model.save("Supervised_LSTM_Model.keras")
print("\nModel saved cleanly as 'Supervised_LSTM_Model.keras'")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout
from tensorflow.keras.utils import Sequence

# =====================================================================
# STEP 1: DEFINE CONFIGURATIONS & SUBSYSTEM CHANNELS
# =====================================================================
TELEMETRY_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/combined_spacecraft_telemetry.csv"
LABELS_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/labels.csv"

# Core structural channels as defined in your team's notebooks
thermal_channels = [f'channel_{i}' for i in list(range(12, 41)) + list(range(47, 53)) + list(range(57, 61))]
adcs_channels = [f'channel_{i}' for i in list(range(53, 57)) + list(range(70, 77))]
target_features = ['datetime'] + thermal_channels + adcs_channels

print("======= STEP 1: LOADING LABELS AND CONFIGURING STREAM =======")
df_labels = pd.read_csv(LABELS_FILE)
df_labels['StartTime'] = pd.to_datetime(df_labels['StartTime']).dt.tz_convert('UTC')
df_labels['EndTime'] = pd.to_datetime(df_labels['EndTime']).dt.tz_convert('UTC')

# =====================================================================
# STEP 2: HIGH-EFFICIENCY CHUNKED LOADING (RAM PROTECTION)
# =====================================================================
# Using a 300,000 row slice to give your model plenty of fault tracking exposure
CHUNK_SIZE = 50000
MAX_ROWS = 300000

X_list = []
y_list = []
rows_processed = 0

print(f"Streaming data sequentially up to {MAX_ROWS} rows...")
for chunk in pd.read_csv(TELEMETRY_FILE, usecols=target_features, chunksize=CHUNK_SIZE):
    if rows_processed >= MAX_ROWS:
        break

    chunk['datetime'] = pd.to_datetime(chunk['datetime'])
    chunk.set_index('datetime', inplace=True)
    if chunk.index.tz is None:
        chunk.index = chunk.index.tz_localize('UTC')

    y_chunk = np.zeros(len(chunk), dtype=np.int32)
    for _, row in df_labels.iterrows():
        mask = (chunk.index >= row['StartTime']) & (chunk.index <= row['EndTime'])
        y_chunk[mask] = 1

    X_list.append(chunk.values.astype(np.float32))
    y_list.append(y_chunk)
    rows_processed += len(chunk)

X_raw = np.vstack(X_list)
y_raw = np.concatenate(y_list)
del X_list, y_list # Instant RAM cleanup

print(f"\nFinal Flat Data Matrix Loaded: {X_raw.shape} (RAM safe!)")
print(f"Class Distribution -> Nominal: {np.sum(y_raw==0)} | Anomalous: {np.sum(y_raw==1)}")

# =====================================================================
# STEP 3: CREATING AN ON-THE-FLY DATA GENERATOR
# =====================================================================
class SpacecraftDataGenerator(Sequence):
    def __init__(self, X_data, y_data, indices, window_size=60, batch_size=256):
        self.X_data = X_data
        self.y_data = y_data
        self.indices = indices  # Keeps track of valid row positions
        self.window_size = window_size
        self.batch_size = batch_size

    def __len__(self):
        return int(np.ceil(len(self.indices) / self.batch_size))

    def __getitem__(self, index):
        # Slice indices for the current batch
        batch_indices = self.indices[index * self.batch_size : (index + 1) * self.batch_size]

        X_batch = []
        y_batch = []

        for idx in batch_indices:
            # Extract sliding temporal slice on the fly without changing global memory allocation
            X_batch.append(self.X_data[idx : idx + self.window_size])
            y_batch.append(self.y_data[idx + self.window_size])

        return np.array(X_batch, dtype=np.float32), np.array(y_batch, dtype=np.int32)

# Define lookback context window
WINDOW_SIZE = 60

# Determine positions where a complete lookback window can be cleanly sliced
valid_start_indices = np.arange(len(X_raw) - WINDOW_SIZE)

# Perform split on index references only (uses negligible bytes of RAM!)
train_idx, test_idx = train_test_split(
    valid_start_indices,
    test_size=0.2,
    random_state=42,
    stratify=y_raw[valid_start_indices + WINDOW_SIZE]
)

# Instantiate our customized memory-safe training streams
train_gen = SpacecraftDataGenerator(X_raw, y_raw, train_idx, window_size=WINDOW_SIZE, batch_size=256)
test_gen = SpacecraftDataGenerator(X_raw, y_raw, test_idx, window_size=WINDOW_SIZE, batch_size=256)

# =====================================================================
# STEP 4: MODEL COMPILATION
# =====================================================================
print("\n======= STEP 4: INITIALIZING LSTM ARCHITECTURE =======")
num_features = X_raw.shape[1]

model = Sequential([
    Input(shape=(WINDOW_SIZE, num_features)),
    LSTM(64, return_sequences=False),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

# =====================================================================
# STEP 5: TRAINING THROUGH DATA GENERATORS
# =====================================================================
print("\n======= STEP 5: TRAINING BASELINE MODEL SUCCESSFULLY =======")
# We train using generators, meaning your RAM usage stays completely flat!
history = model.fit(
    train_gen,
    epochs=6,
    validation_data=test_gen,
    verbose=1
)

# =====================================================================
# STEP 6: EVALUATION METRICS
# =====================================================================
print("\n======= STEP 6: GENERATING PERFORMANCE METRICS =======")

# Gather predictions step-by-step from our streaming generator
y_true_list = []
y_pred_list = []

for i in range(len(test_gen)):
    bx, by = test_gen[i]
    b_preds = model.predict_on_batch(bx)
    y_true_list.extend(by)
    y_pred_list.extend(b_preds.flatten())

y_true = np.array(y_true_list)
y_pred = (np.array(y_pred_list) > 0.5).astype(int)

# Extract True Metric Layout
acc = accuracy_score(y_true, y_pred)
prec = precision_score(y_true, y_pred, zero_division=0)
rec = recall_score(y_true, y_pred, zero_division=0)
f1 = f1_score(y_true, y_pred, zero_division=0)

print("\n================ FINAL MODEL 1 METRICS ================")
print(f"Accuracy Performance : {acc * 100:.2f}%")
print(f"Precision Score      : {prec:.4f}")
print(f"Recall (Sensitivity) : {rec:.4f}")
print(f"F1-Score Performance : {f1:.4f}")
print("\nConfusion Matrix Output:")
print(confusion_matrix(y_true, y_pred))
print("\nDetailed Summary Breakdown:")
print(classification_report(y_true, y_pred, zero_division=0))

# Save using native keras layout
model.save("Supervised_LSTM_Model.keras")
print("\nModel saved cleanly as 'Supervised_LSTM_Model.keras'")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
import tensorflow as tf
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Input, Conv1D, MaxPooling1D, Flatten, Dense, Dropout, Layer
import tensorflow.keras.backend as K
from tensorflow.keras.utils import Sequence

# =====================================================================
# STEP 1: LOAD PREPROCESSED DATA STREAM (RE-USE BALANCED MEMORY)
# =====================================================================
TELEMETRY_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/combined_spacecraft_telemetry.csv"
LABELS_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/labels.csv"

thermal_channels = [f'channel_{i}' for i in list(range(12, 41)) + list(range(47, 53)) + list(range(57, 61))]
adcs_channels = [f'channel_{i}' for i in list(range(53, 57)) + list(range(70, 77))]
target_features = ['datetime'] + thermal_channels + adcs_channels

print("======= STEP 1: INITIALIZING STREAMS & COMPUTING BALANCED WEIGHTS =======")
df_labels = pd.read_csv(LABELS_FILE)
df_labels['StartTime'] = pd.to_datetime(df_labels['StartTime']).dt.tz_convert('UTC')
df_labels['EndTime'] = pd.to_datetime(df_labels['EndTime']).dt.tz_convert('UTC')

CHUNK_SIZE = 50000
MAX_ROWS = 300000
X_list, y_list = [], []
rows_processed = 0

for chunk in pd.read_csv(TELEMETRY_FILE, usecols=target_features, chunksize=CHUNK_SIZE):
    if rows_processed >= MAX_ROWS:
        break
    chunk['datetime'] = pd.to_datetime(chunk['datetime'])
    chunk.set_index('datetime', inplace=True)
    if chunk.index.tz is None: chunk.index = chunk.index.tz_localize('UTC')

    y_chunk = np.zeros(len(chunk), dtype=np.int32)
    for _, row in df_labels.iterrows():
        mask = (chunk.index >= row['StartTime']) & (chunk.index <= row['EndTime'])
        y_chunk[mask] = 1

    X_list.append(chunk.values.astype(np.float32))
    y_list.append(y_chunk)
    rows_processed += len(chunk)

X_raw = np.vstack(X_list)
y_raw = np.concatenate(y_list)
del X_list, y_list

# Calculate balancing mathematical class weights (Nominal vs Anomaly penalty scaling)
neg_count = np.sum(y_raw == 0)
pos_count = np.sum(y_raw == 1)
total = neg_count + pos_count

weight_for_0 = (1 / neg_count) * (total / 2.0)
weight_for_1 = (1 / pos_count) * (total / 2.0)
class_weights = {0: weight_for_0, 1: weight_for_1}
print(f"Computed Imbalance Penalties -> Class 0: {weight_for_0:.2f} | Class 1: {weight_for_1:.2f}")

# =====================================================================
# STEP 2: ON-THE-FLY GENERATOR CONFIGURATION
# =====================================================================
class SpacecraftDataGenerator(Sequence):
    def __init__(self, X_data, y_data, indices, window_size=60, batch_size=256):
        self.X_data = X_data
        self.y_data = y_data
        self.indices = indices
        self.window_size = window_size
        self.batch_size = batch_size

    def __len__(self):
        return int(np.ceil(len(self.indices) / self.batch_size))

    def __getitem__(self, index):
        batch_indices = self.indices[index * self.batch_size : (index + 1) * self.batch_size]
        X_batch, y_batch = [], []
        for idx in batch_indices:
            X_batch.append(self.X_data[idx : idx + self.window_size])
            y_batch.append(self.y_data[idx + self.window_size])
        return np.array(X_batch, dtype=np.float32), np.array(y_batch, dtype=np.int32)

WINDOW_SIZE = 60
valid_start_indices = np.arange(len(X_raw) - WINDOW_SIZE)
train_idx, test_idx = train_test_split(
    valid_start_indices, test_size=0.2, random_state=42,
    stratify=y_raw[valid_start_indices + WINDOW_SIZE]
)

train_gen = SpacecraftDataGenerator(X_raw, y_raw, train_idx, window_size=WINDOW_SIZE, batch_size=256)
test_gen = SpacecraftDataGenerator(X_raw, y_raw, test_idx, window_size=WINDOW_SIZE, batch_size=256)

# =====================================================================
# STEP 3: CUSTOM ATTTENTION LAYER ARCHITECTURE FOR SPACECRAFT CHANNELS
# =====================================================================
class SpatialTemporalAttention(Layer):
    def __init__(self, **kwargs):
        super(SpatialTemporalAttention, self).__init__(**kwargs)

    def build(self, input_shape):
        self.W = self.add_weight(name='att_weight', shape=(input_shape[-1], 1), initializer='glorot_uniform', trainable=True)
        self.b = self.add_weight(name='att_bias', shape=(input_shape[1], 1), initializer='zeros', trainable=True)
        super(SpatialTemporalAttention, self).build(input_shape)

    def call(self, x):
        e = K.tanh(K.dot(x, self.W) + self.b)
        e = K.squeeze(e, axis=-1)
        alpha = K.softmax(e)
        alpha = K.expand_dims(alpha, axis=-1)
        return x * alpha

# =====================================================================
# STEP 4: MODEL 2 DESIGN - HYBRID CNN + ATTENTION CLASSIFIER
# =====================================================================
print("\n======= STEP 4: INITIALIZING HYBRID CNN + ATTENTION ARCHITECTURE =======")
num_features = X_raw.shape[1]

inputs = Input(shape=(WINDOW_SIZE, num_features))
x = Conv1D(filters=64, kernel_size=3, activation='relu', padding='same')(inputs)
x = MaxPooling1D(pool_size=2)(x)
x = SpatialTemporalAttention()(x) # Dynamically weights cross-channel heat changes
x = Flatten()(x)
x = Dense(32, activation='relu')(x)
x = Dropout(0.3)(x)
outputs = Dense(1, activation='sigmoid')(x)

hybrid_model = Model(inputs, outputs)
hybrid_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
hybrid_model.summary()

# =====================================================================
# STEP 5: HYBRID NETWORK BALANCED TRAINING
# =====================================================================
print("\n======= STEP 5: TRAINING HYBRID MODEL WITH LOSS BALANCING =======")
history_hybrid = hybrid_model.fit(
    train_gen,
    epochs=6,
    validation_data=test_gen,
    class_weight=class_weights, # Overcomes the 0-prediction shortcut trick completely
    verbose=1
)

# =====================================================================
# STEP 6: PERFORMANCE EVALUATION & SUMMARY MATRICES
# =====================================================================
print("\n======= STEP 6: COMPUTING PERFORMANCE METRICS =======")
y_true_list, y_pred_list = [], []
for i in range(len(test_gen)):
    bx, by = test_gen[i]
    b_preds = hybrid_model.predict_on_batch(bx)
    y_true_list.extend(by)
    y_pred_list.extend(b_preds.flatten())

y_true = np.array(y_true_list)
y_pred = (np.array(y_pred_list) > 0.5).astype(int)

acc = accuracy_score(y_true, y_pred)
prec = precision_score(y_true, y_pred, zero_division=0)
rec = recall_score(y_true, y_pred, zero_division=0)
f1 = f1_score(y_true, y_pred, zero_division=0)

print("\n================ FINAL HYBRID MODEL 2 METRICS ================")
print(f"Hybrid Accuracy : {acc * 100:.2f}%")
print(f"Hybrid Precision: {prec:.4f}")
print(f"Hybrid Recall   : {rec:.4f}")
print(f"Hybrid F1-Score : {f1:.4f}")
print("\nTrue Balanced Confusion Matrix:")
print(confusion_matrix(y_true, y_pred))
print("\nClassification Summary Report:")
print(classification_report(y_true, y_pred, zero_division=0))

# Save Native Keras Model Structure
hybrid_model.save("Hybrid_CNN_Attention_Model.keras")
print("\nHybrid model file saved successfully as 'Hybrid_CNN_Attention_Model.keras'")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout, Conv1D, MaxPooling1D, Flatten, Layer
import tensorflow.keras.backend as K
from tensorflow.keras.utils import Sequence
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# =====================================================================
# STEP 1: LOAD RE-USABLE DATA STREAM & COMPUTE BALANCED WEIGHTS
# =====================================================================
TELEMETRY_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/combined_spacecraft_telemetry.csv"
LABELS_FILE = "/content/drive/MyDrive/Spacecraft_Project_Data/labels.csv"

thermal_channels = [f'channel_{i}' for i in list(range(12, 41)) + list(range(47, 53)) + list(range(57, 61))]
adcs_channels = [f'channel_{i}' for i in list(range(53, 57)) + list(range(70, 77))]
target_features = ['datetime'] + thermal_channels + adcs_channels

print("======= STEP 1: INITIALIZING STREAMS & COMPUTING BALANCED WEIGHTS =======")
df_labels = pd.read_csv(LABELS_FILE)
df_labels['StartTime'] = pd.to_datetime(df_labels['StartTime']).dt.tz_convert('UTC')
df_labels['EndTime'] = pd.to_datetime(df_labels['EndTime']).dt.tz_convert('UTC')

CHUNK_SIZE = 50000
MAX_ROWS = 300000
X_list, y_list = [], []
rows_processed = 0

for chunk in pd.read_csv(TELEMETRY_FILE, usecols=target_features, chunksize=CHUNK_SIZE):
    if rows_processed >= MAX_ROWS:
        break
    chunk['datetime'] = pd.to_datetime(chunk['datetime'])
    chunk.set_index('datetime', inplace=True)
    if chunk.index.tz is None: chunk.index = chunk.index.tz_localize('UTC')

    y_chunk = np.zeros(len(chunk), dtype=np.int32)
    for _, row in df_labels.iterrows():
        mask = (chunk.index >= row['StartTime']) & (chunk.index <= row['EndTime'])
        y_chunk[mask] = 1

    X_list.append(chunk.values.astype(np.float32))
    y_list.append(y_chunk)
    rows_processed += len(chunk)

X_raw = np.vstack(X_list)
y_raw = np.concatenate(y_list)
del X_list, y_list

# Class weights configuration to prevent "predict all zero" shortcuts
neg_count = np.sum(y_raw == 0)
pos_count = np.sum(y_raw == 1)
total = neg_count + pos_count
class_weights = {0: (1 / neg_count) * (total / 2.0), 1: (1 / pos_count) * (total / 2.0)}

# =====================================================================
# STEP 2: DATA GENERATOR SETUP
# =====================================================================
class SpacecraftDataGenerator(Sequence):
    def __init__(self, X_data, y_data, indices, window_size=60, batch_size=256):
        self.X_data = X_data
        self.y_data = y_data
        self.indices = indices
        self.window_size = window_size
        self.batch_size = batch_size

    def __len__(self):
        return int(np.ceil(len(self.indices) / self.batch_size))

    def __getitem__(self, index):
        batch_indices = self.indices[index * self.batch_size : (index + 1) * self.batch_size]
        X_batch, y_batch = [], []
        for idx in batch_indices:
            X_batch.append(self.X_data[idx : idx + self.window_size])
            y_batch.append(self.y_data[idx + self.window_size])
        return np.array(X_batch, dtype=np.float32), np.array(y_batch, dtype=np.int32)

WINDOW_SIZE = 60
valid_start_indices = np.arange(len(X_raw) - WINDOW_SIZE)
train_idx, test_idx = train_test_split(
    valid_start_indices, test_size=0.2, random_state=42,
    stratify=y_raw[valid_start_indices + WINDOW_SIZE]
)

train_gen = SpacecraftDataGenerator(X_raw, y_raw, train_idx, window_size=WINDOW_SIZE, batch_size=256)
test_gen = SpacecraftDataGenerator(X_raw, y_raw, test_idx, window_size=WINDOW_SIZE, batch_size=256)
num_features = X_raw.shape[1]

# =====================================================================
# STEP 3: CUSTOM ATTENTION LAYER DEFINITION
# =====================================================================
class SpatialTemporalAttention(Layer):
    def __init__(self, **kwargs):
        super(SpatialTemporalAttention, self).__init__(**kwargs)

    def build(self, input_shape):
        self.W = self.add_weight(name='att_weight', shape=(input_shape[-1], 1), initializer='glorot_uniform', trainable=True)
        self.b = self.add_weight(name='att_bias', shape=(input_shape[1], 1), initializer='zeros', trainable=True)
        super(SpatialTemporalAttention, self).build(input_shape)

    def call(self, x):
        e = K.tanh(K.dot(x, self.W) + self.b)
        e = K.squeeze(e, axis=-1)
        alpha = K.softmax(e)
        alpha = K.expand_dims(alpha, axis=-1)
        return x * alpha

# =====================================================================
# STEP 4: TRAIN MODEL 1 (BASELINE LSTM)
# =====================================================================
print("\n======= TRAINING MODEL 1: BASELINE LSTM =======")
inputs_1 = Input(shape=(WINDOW_SIZE, num_features))
x1 = LSTM(64, return_sequences=False)(inputs_1)
x1 = Dropout(0.3)(x1)
x1 = Dense(32, activation='relu')(x1)
outputs_1 = Dense(1, activation='sigmoid')(x1)

model_lstm = Model(inputs_1, outputs_1)
model_lstm.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_lstm = model_lstm.fit(train_gen, epochs=5, validation_data=test_gen, class_weight=class_weights, verbose=1)

# =====================================================================
# STEP 5: TRAIN MODEL 2 (HYBRID CNN + ATTENTION)
# =====================================================================
print("\n======= TRAINING MODEL 2: HYBRID CNN + ATTENTION =======")
inputs_2 = Input(shape=(WINDOW_SIZE, num_features))
x2 = Conv1D(filters=64, kernel_size=3, activation='relu', padding='same')(inputs_2)
x2 = MaxPooling1D(pool_size=2)(x2)
x2 = SpatialTemporalAttention()(x2)
x2 = Flatten()(x2)
x2 = Dense(32, activation='relu')(x2)
x2 = Dropout(0.3)(x2)
outputs_2 = Dense(1, activation='sigmoid')(x2)

model_hybrid = Model(inputs_2, outputs_2)
model_hybrid.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

history_hybrid = model_hybrid.fit(train_gen, epochs=5, validation_data=test_gen, class_weight=class_weights, verbose=1)

# =====================================================================
# STEP 6: EVALUATE BOTH MODELS ON THE SAME EVALUATION PIPELINE
# =====================================================================
print("\n======= STEP 6: COMPUTING SIDE-BY-SIDE METRICS =======")

# Gather Ground Truths and Predictions
y_true = []
preds_lstm = []
preds_hybrid = []

for i in range(len(test_gen)):
    bx, by = test_gen[i]
    y_true.extend(by)
    preds_lstm.extend(model_lstm.predict_on_batch(bx).flatten())
    preds_hybrid.extend(model_hybrid.predict_on_batch(bx).flatten())

y_true = np.array(y_true)
y_pred_lstm = (np.array(preds_lstm) > 0.5).astype(int)
y_pred_hybrid = (np.array(preds_hybrid) > 0.5).astype(int)

# Calculate Statistical Performance Summaries
metrics = {
    "Metric": ["Accuracy", "Precision", "Recall", "F1-Score"],
    "Model 1 (Baseline LSTM)": [
        accuracy_score(y_true, y_pred_lstm),
        precision_score(y_true, y_pred_lstm, zero_division=0),
        recall_score(y_true, y_pred_lstm, zero_division=0),
        f1_score(y_true, y_pred_lstm, zero_division=0)
    ],
    "Model 2 (Hybrid CNN+Attention)": [
        accuracy_score(y_true, y_pred_hybrid),
        precision_score(y_true, y_pred_hybrid, zero_division=0),
        recall_score(y_true, y_pred_hybrid, zero_division=0),
        f1_score(y_true, y_pred_hybrid, zero_division=0)
    ]
}

df_comparison = pd.DataFrame(metrics)
print("\n==================== PROJECT PERFORMANCE RESULT MATRIX ====================")
print(df_comparison.to_string(index=False))
print("===========================================================================")

print("\nConfusion Matrix - Model 1 (LSTM):")
print(confusion_matrix(y_true, y_pred_lstm))

print("\nConfusion Matrix - Model 2 (CNN + Attention):")
print(confusion_matrix(y_true, y_pred_hybrid))

# =====================================================================
# STEP 7: RENDER COMPARATIVE PLOTS FOR REPORT VISUALS
# =====================================================================
epochs_range = range(1, 6)

plt.figure(figsize=(12, 5))

# Plot Comparative Loss curves
plt.subplot(1, 2, 1)
plt.plot(epochs_range, history_lstm.history['val_loss'], 'b-o', label='Model 1: LSTM Val Loss')
plt.plot(epochs_range, history_hybrid.history['val_loss'], 'r-s', label='Model 2: Hybrid Val Loss')
plt.title('Validation Loss Comparison')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

# Plot Comparative Accuracy curves
plt.subplot(1, 2, 2)
plt.plot(epochs_range, history_lstm.history['val_accuracy'], 'b-o', label='Model 1: LSTM Val Acc')
plt.plot(epochs_range, history_hybrid.history['val_accuracy'], 'r-s', label='Model 2: Hybrid Val Acc')
plt.title('Validation Accuracy Comparison')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
